# Business Understanding

## Business Goal

O objetivo da Pergunta 1 é prever a classificação final das equipas numa época antes dessa época começar.

A previsão deve utilizar apenas informação disponível até ao início da época alvo, evitando qualquer utilização de informação futura.

## Data Mining Goal

O problema é tratado como um problema de regressão.

O modelo prevê o desempenho esperado de cada equipa em pontos por jogo e, posteriormente, essas previsões são usadas para ordenar as equipas dentro de cada divisão e obter o ranking previsto.

A métrica mais importante é o Mean Rank Error, porque mede diretamente a diferença média entre a posição prevista e a posição real.

# Data Understanding

Os dados utilizados vêm de diferentes tabelas relacionadas com equipas, jogadores, guarda-redes e treinadores.

As principais fontes utilizadas são:

- `Teams`: resultados e classificação das equipas por época;
- `Scoring`: estatísticas ofensivas dos jogadores;
- `Goalies`: estatísticas dos guarda-redes;
- `Coaches`: informação sobre os treinadores de cada equipa.

Os dados abrangem várias épocas, o que torna importante considerar a componente temporal do problema.

Foram analisadas variáveis relacionadas com o desempenho das equipas, como pontos, golos marcados, golos sofridos, desempenho dos jogadores, percentagem de defesas e número de treinadores.

Como algumas épocas têm números diferentes de jogos, várias estatísticas foram transformadas em valores por jogo, permitindo comparações mais consistentes entre épocas.

Também existem valores em falta em algumas variáveis históricas, que são tratados posteriormente durante a preparação dos dados.

A análise dos dados mostrou ainda que o ranking presente no dataset corresponde ao ranking dentro da divisão, e não a uma classificação global da liga. Por esse motivo, as previsões finais são também avaliadas dentro de cada divisão.

# Data Quality Considerations

Durante a preparação dos dados foram considerados alguns problemas de qualidade, nomeadamente valores em falta, diferenças no número de jogos entre épocas e necessidade de converter corretamente algumas variáveis para formato numérico.

Estes aspetos são importantes para garantir que as comparações entre equipas e épocas são consistentes.

# Temporal Considerations

Como os dados representam várias épocas consecutivas, a ordem temporal deve ser respeitada.

Para prever uma determinada época, apenas são utilizados dados de épocas anteriores, evitando que informação futura influencie o treino do modelo.

# Interpretation of Results

A análise dos resultados permitiu identificar quais os modelos que apresentam menor erro de previsão e quais as variáveis históricas que parecem ser mais úteis para prever o desempenho futuro das equipas.

Foi também possível observar que o desempenho dos modelos varia entre épocas, mostrando que algumas épocas são mais difíceis de prever do que outras.

# Visual Analysis

Os gráficos são utilizados para comparar o desempenho dos diferentes modelos e para analisar a evolução do erro ao longo das épocas.

Estas visualizações permitem identificar de forma mais clara diferenças entre algoritmos, variações temporais e situações em que o modelo apresenta melhor ou pior desempenho.
