# Question 1 - Final Regular Season Ranking

## Business Understanding

The goal is to predict the final regular season ranking of the teams
for a target season.

The prediction is assumed to be made one day before the first game
of the new season.

Therefore, only information available before the beginning of the
target season can be used.

The output should be an ordering of the teams according to their
predicted final performance.

## Initial modelling idea

This will initially be treated as a regression problem.

A numerical measure of team performance will be predicted for each team,
and teams will then be ordered according to the predicted value.
